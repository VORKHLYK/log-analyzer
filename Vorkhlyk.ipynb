```python
%%writefile log.txt
08:01 success 10.0.0.5 liliia
08:02 failed 45.13.7.9 admin
08:02 failed 45.13.7.9 admin
08:03 failed 45.13.7.9 root
08:03 success 10.0.0.8 tom
08:05 failed 10.0.0.5 liliia
```

```python
counts = {}

with open("log.txt") as f:
  for line in f:
    parts = line.strip().split()
    status = parts[1]
    ip = parts[2]

    if status == "failed":
      counts[ip] = counts.get(ip,0)+1
for ip, n in counts.items():
  if n>=3:
    print ("Flagged IP:", ip, "- attempts:", n)
```
